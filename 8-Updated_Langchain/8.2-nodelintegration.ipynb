### Streaming
Most models can stream their output content while it is being genrated By displaying output progressively, streaming significantly improves user exprence, particularly for longer response. Calling Stream() returns an iterator that yelds output chinks as they are produce . You can use a loop to process each chunks in real time:

```python
model.invoke("write me a 200 words paragraph an Artificial Intelligence")
```

```python
response=model.stream("Why do parrots have colorful feathers?")
```

```python
for chunk in model.stream("Why do parrots have colorful feathers?"):
    print(chunk.text, end=" ",flush=True)
```
