```python
from langchain_openai import ChatOpenAI
from langchain.agents import create_tool_calling_agent, AgentExecutor
from langchain_core.prompts import ChatPromptTemplate

# 1. Initialize the Brain (GPT-5)
llm = ChatOpenAI(model="gpt-5", temperature=0)

# 2. Define the Prompt (Mandatory for Agents)
# The placeholders {input} and {agent_scratchpad} are required
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant."),
    ("human", "{input}"),
    ("placeholder", "{agent_scratchpad}"),
])

# 3. Create the Agent Logic
# Use create_tool_calling_agent instead of create_agent
agent = create_tool_calling_agent(llm, tools=[], prompt=prompt)

# 4. Create the Executor (The runtime)
agent_executor = AgentExecutor(agent=agent, tools=[], verbose=True)

# Example call:
# agent_executor.invoke({"input": "What is the capital of France?"})
```
