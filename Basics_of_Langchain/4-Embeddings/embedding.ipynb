### Ollama 
Ollama supports embedding models making it possible to build retrival augmented generation(RAG) application that combine text prompts with exiting documents or other data.

```python
from langchain_community.embeddings import OllamaEmbeddings
```

```python
embedding=(
    OllamaEmbeddings(model="gemma:2b") #
)
```

```python
r1=embedding.embed_documents(
    [
        "Alpha is the frist letter of Greek alphabet",
        "Beta is the Second letter of Greek alphabet"
    ]
)
```
